```python
import requests
import json

# The API endpoint URL
url = "http://localhost:11434/api/generate"

# The data to send (as a Python dictionary)
new_post_data = {
  "model": "gemma3:4b",
  "prompt": "Why is the sky blue?"
}

# Make the GET request to the API
response = requests.get(url,json=new_post_data)

print(str(response))

```
