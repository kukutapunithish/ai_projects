```python
import requests
import json

# The API endpoint URL
url = "http://localhost:11434/api/generate"

# The data to send (as a Python dictionary)
new_post_data = {
  "model": "qwen3.5:2b",
  "prompt": "Why is the sky blue?"
}

# Make the GET request to the API
response = requests.get(url,json=new_post_data)

print(str(response))

```

```python
res_con = response.content
```

```python
import json
import base64

def convert_bytes_to_json(bytes_data):
    """Convert bytes to JSON format with error handling"""
    try:
        # Decode bytes to string
        if isinstance(bytes_data, bytes):
            json_string = json.dumps(bytes_data.decode('utf-8'))
        else:
            json_string = json.dumps(bytes_data)
        
        # Parse JSON string back to bytes
        parsed_bytes = json.loads(json_string)
        return parsed_bytes
    except Exception as e:
        print(f"Error: {e}")
        return None

result = convert_bytes_to_json(res_con)
print(result)
```
