```python
import json

def get_all_keys(obj):
    keys = set()
    if isinstance(obj, dict):
        for k, v in obj.items():
            keys.add(k)
            keys |= get_all_keys(v)
    elif isinstance(obj, list):
        for item in obj:
            keys |= get_all_keys(item)
    return keys

# Example usage
data = json.loads(documents[1].to_json())
all_keys = get_all_keys(data)
print(all_keys)
```
